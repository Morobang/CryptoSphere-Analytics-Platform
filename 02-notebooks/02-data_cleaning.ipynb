# Data Cleaning Notebook

## Purpose
This notebook handles data cleaning and preprocessing tasks including:
- Data type standardization
- Handling missing values
- Removing duplicates
- Outlier detection and treatment
- Data validation and quality checks

## Process Flow
1. **Load Raw Data** - Import data from bronze layer
2. **Data Profiling** - Understand data characteristics
3. **Missing Value Treatment** - Handle nulls and missing data
4. **Data Type Conversion** - Standardize data types
5. **Duplicate Removal** - Identify and remove duplicates
6. **Outlier Detection** - Find and treat outliers
7. **Data Validation** - Ensure data quality standards
8. **Export Clean Data** - Save to processed layer
